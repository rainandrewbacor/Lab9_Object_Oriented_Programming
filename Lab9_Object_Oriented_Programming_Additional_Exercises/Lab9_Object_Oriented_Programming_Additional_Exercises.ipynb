```python
class LibraryManagementSystem:
    def __init__(self, number):
        self.number = number
        self.__book_list = []
        self.__borrow_list = []
        print(f"Library Management System Initialized! Books to be inputted: {number}")
        
    def check_num(self):
        if len(self.__book_list) == 0:
            return "[WARNING] There is no books in the library list. Please re-run if this is not intended."
        if self.number != len(self.__book_list):
            raise ValueError(f"Book input only contains {len(self.__book_list)}, while it should be {self.number}")
        return "[INFO] Book number in initialization passes the number of inputted books."
        
    def display_book(self):
        if not self.__book_list:
            return f"[WARNING] No books are added in the library yet."
        output = ["\n---- LIBRARY BOOK LIST ---"]
        for book in self.__book_list:
            status = "Available" if book[2] else "Borrowed"
            output.append(f"Title: {book[0]:<40} | Author: {book[1]:<25} | Availability: {status:<15}")
        return "\n".join(output) + "\n"

    def add_book(self, title: str, author: str, avail_stats: bool):
        if not isinstance(title, str) or title.isdigit():
            return f"[WARNING] Title must be a proper string."
        title = title.strip()
        self.__book_list.append([title, author, avail_stats])
        return f"[INFO] Added book with the following details:\nTitle: {title}\nAuthor: {author}\nAvailability Status: {avail_stats}\n{self.display_book()}"

    def borrow_book(self, title: str):
        if not isinstance(title, str) or title.isdigit():
            return f"[WARNING] Title must be a proper string."
        for book in self.__book_list:     
            if book[0] == title:
                if not book[2]:
                    return f"[WARNING] {book} is already borrowed."
                book[2] = False
                self.__borrow_list.append(title)
                return f"[INFO] The book entitled: '{title}' has been borrowed. Make sure to return it in time!\n{self.display_book()}"
        return f"[WARNING] The book appears to not be on the library list. Make sure to add them first."

    def return_book(self, title: str):
        if not isinstance(title, str) or title.isdigit():
            return f"[WARNING] Title must be a proper string."
        if title not in self.__borrow_list:
            return f"[WARNING] The book title does not appear on the borrowed list. Did you borrow them first?"

        for book in self.__book_list:
            if book[0] == title:
                book[2] = True
                self.__borrow_list.remove(title)
                return f"[INFO] The book entitled: '{title} has been returned.\n{self.display_book()}"
        
try:
    num_input = int(input("How many books do you want to input?"))
    library = LibraryManagementSystem(num_input)
except ValueError as exce:
    raise ValueError(f"Please only input numbers.")

def lib_prompt():
    session_input = input("Would you like to start a session (yes/no).")    
    if session_input == "yes":      
        action_input = input("What would you like to do? (add/return/borrow)")
        if action_input == "add":
            title = input("What will be the title of the book?")
            author = input("Who will be the author of the book?")
            avail = input("Will this book be available (True)? or borrowed (False)?").strip().lower()
            if avail in ("true", "false"):
                avail_bool = (avail == "true")
                print(library.add_book(title, author, avail_bool))
            else:
                print("[WARNING] Please only enter True or False.")
        elif action_input == "borrow":
            title = input("What will be the title of the book?")
            print(library.borrow_book(title))
        elif action_input == "return":
            title = input("What will be the title of the book?")
            print(library.return_book(title))
        else:
            print("Please only input valid input (add/return/borrow).")
        return False
    elif session_input == "no":
        print("Session done. Thank you!")
        print(library.check_num())
        return True
    else: 
        print("Please only input valid input (yes/no)")
        return False
    
done_session = False
while not done_session:
    done_session = lib_prompt()


        
```

```python
STUDENT_ID = "TUPM-26-1879"
id_digits = "".join(d for d in STUDENT_ID if d.isdigit())
id_sum = sum(map(int, id_digits))
hourly_wage = (id_sum * 200) + 50

class EmployeePayrollSystem:
    def __init__(self, number, hour_wage):
        if hour_wage < 0:
            raise ValueError(f"Hourly wage cannot be negative.")
        self.hour_wage = hour_wage   
        self.number = number
        self.__employee_list = []

    def check_num(self):
        if len(self.__employee_list) == 0:
            return "[WARNING] There is no employees in the employee list. Please re-run if this is not intended."
        if self.number != len(self.__employee_list):
            raise ValueError(f"Employee input only contains {len(self.__employee_list)}, while it should be {self.number}")
        return "[INFO] Employee number in initialization passes the number of inputted employee."
    
    def display_employ(self):
        if not self.__employee_list:
            return f"[WARNING] No employees are added in the system yet."
        output = [f"\n=============== EMPLOYEE PAYROLL LIST - HOURLY WAGE: {self.hour_wage} ==============="]
        for employ in self.__employee_list:
            status = "Yes" if employ[2] > 160 else "No"
            output.append(f"Name: {employ[0]:<20} | Hours Worked: {employ[2]:<15} | Overtime?: {status:<13} | Overtime Hours: {max(0, employ[2] - 160):<15} | Salary: {employ[3]:.2f}")
        return "\n".join(output) + "\n"
    
    def add_employee(self, name, hour_worked):
        if hour_worked < 0:
            return f"The system cannot accept negative integers. Enter valid work hours."
        self.__employee_list.append([name, self.hour_wage, hour_worked, 0.0])
        return f"Added Employee: {name}, Hours Worked: {hour_worked}"

    def compute_salary(self):
        if not self.__employee_list:
            return f"No employee has been added in the system. Ensure at least one employee entered."
        
        for employ in self.__employee_list:
            if employ[2] > 160:
                base_salary = employ[1] * 160
                rate = employ[1] * 1.5
                xtra_hrs = employ[2] - 160
                ot_extra = xtra_hrs * rate 
                ot_salary = base_salary + ot_extra
                total_salary = ot_salary
            else:
                base_salary = employ[1] * employ[2]
                total_salary = base_salary
            employ[3] = total_salary
        return f"{self.display_employ()}"

try:
    num_input = int(input("How many employees do you wish to input?"))
    employ = EmployeePayrollSystem(num_input, hourly_wage)
except ValueError as exce:
    raise ValueError(f"Please only input numbers.")

def session_prompt():
    done_input = input("Welcome. Do you want to continue this session? (yes/no)").strip().lower()
    if done_input == "yes":
        name_input = input("Please state the employee name you want to add: ").strip()
        try:
            hour_input = int(input("State their work hours: "))
            print(employ.add_employee(name_input, hour_input))
        except ValueError:
            print(f"Please only input numbers on employee's work hours")
        return False
    elif done_input == "no":
        print(employ.compute_salary())
        print("Finalized computation. Thank you for utilising this tool!")
        print(employ.check_num())
        return True
    else:
        print("Invalid input. Please only input yes or no.")
        return False

done_compute = False
while not done_compute:
    done_compute = session_prompt()



        
        




    
```

```python
product_list = [
    {
        "product": "shampoo",
        "price": 10.00,
        "stock": 50
    },
    {
        "product": "soap",
        "price": 25.00,
        "stock": 10
    },
    {
        "product": "toothpaste",
        "price": 50.00,
        "stock": 0
    },
    {
        "product": "mani",
        "price": 20.00,
        "stock": 6
    },
    {
        "product": "laptop",
        "price": 10000.00,
        "stock": 1
    }
]

class OnlineShoppingCartSystem:
    def __init__(self):
        self.__shopping_cart = []

    def display_list(self):
        output = [f"\n===============  PRODUCT LIST  ==============="]
        for item in product_list:
            content = f"Product: {item['product']:<13} | Price: {item['price']:<15} | Stock: {item['stock']:.2f}"
            output.append(content)
        return "\n".join(output) + "\n"

    def add_items(self, name, qtty):        
        if not (isinstance(qtty, int) and isinstance(name, str)):
            return f"Please input valid product name and quantity." 
        if qtty <= 0:
            return f"Quantity must be greater than zero."    
        for prod in product_list:
            if prod["product"] == name:
                if qtty <= prod["stock"]:
                    prod["stock"] = prod["stock"] - qtty
                    total_cost = qtty * prod["price"]

                    for item in self.__shopping_cart:
                        if item[0] == name:
                            item[1] += qtty
                            item[2] += total_cost
                            return f"Updated {name} - {qtty}. Cost: {total_cost}"                        
                    self.__shopping_cart.append([name, qtty, total_cost])
                    return f"Added {name} - {qtty}. Cost: {total_cost}\n{self.display_list()}"
                elif prod["stock"] == 0:
                    return f"Unfortunately, {name} is out of stock."
                elif qtty > prod["stock"]:
                    return f"You cannot add more than the available stock."
        return f"The product '{name} is not on the product list. Please select from the available products only."

    def remove_items(self, name, qtty):
        if qtty <= 0:
            return f"Quantity must be greater than zero."
        
        for item in self.__shopping_cart:
            if item[0] == name:
                if qtty > item[1]:
                    return f"Cannot remote {qtty} from {name} as it exceeds the available item count in the shopping cart."
                
                for prod in product_list:
                    if prod["product"] == name:
                        prod["stock"] += qtty
                        item[1] -= qtty
                        total_cost = item[1] * prod["price"]
                        item[2] = total_cost

                        if item[1] == 0:
                            self.__shopping_cart.remove(item)
                            return f"Removed {item[0]} from your cart."
                    
                        return f"Removed {name} - {qtty}. Cost: {total_cost}"
                
        return f"The product '{name}' is not on the product list. Please select from the available products only."

    def display_price(self):
        if not self.__shopping_cart:
            return f"Shopping Cart is empty."
        total_cost = 0
        output = [f"\n===============  SHOPPING CART  ==============="]
        for item in self.__shopping_cart:
            content = f"Product: {item[0]:<13} | Quantity: {item[1]:<15} | Cost: {item[2]:.2f}"
            output.append(content)
            total_cost += item[2]
        output.append(f"{'-' * len(content)}")
        output.append(f"Total Cost: P{total_cost:.2f}")
        return "\n".join(output) + "\n"
                    
# USER PROMPTING
shopping = OnlineShoppingCartSystem()
output = [f"\n===============  PRODUCT LIST  ==============="]
for item in product_list:
    content = f"Product: {item['product']:<13} | Price: {item['price']:<15} | Stock: {item['stock']:.2f}"
    output.append(content)
print(f"{'\n'.join(output) + '\n'}")
def user_prompt():
    done_input = input("Welcome. Do you want to continue this session? (yes/no)").strip().lower()
    if done_input == "yes":
        action_input = input("What action would you like to do? (add/remove): ")
        if action_input == "add":
            name_input = input("Please state the product name you want to add: ").strip().lower()
            try:
                qtty_input = int(input("How many would you like to add?: "))
                print(shopping.add_items(name_input, qtty_input))
            except ValueError:
                print("Please only input valid numbers as quantity.")
        elif action_input == "remove":
            name_input = input("Please state the product name you want to remove: ").strip().lower()
            try:
                qtty_input = int(input("How many would you like to remove?: "))
                print(shopping.remove_items(name_input, qtty_input))
            except ValueError:
                print("Please only input valid numbers as quantity.")
        return False
    elif done_input == "no":
        print("Finalized Shopping Cart. Thank You for Shopping!")
        print(shopping.display_price())
        return True
    else:
        print("Invalid input. Please only input yes or no.")
        return False

done_shopping = False
while not done_shopping:
    done_shopping = user_prompt()
    


            

    
```
